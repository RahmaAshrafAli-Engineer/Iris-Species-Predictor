# Iris Flower category prediction using Logistic Regression 🪻

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_iris
import pandas as pd

# 1. Load the data
iris = load_iris()

# تحويلها إلى DataFrame ليسهل التعامل معها
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df['target'] = iris.target

# عرض أول 5 صفوف
iris_df.head()
```

```python
iris_df.info()
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# 2. Load and split the data
iris = load_iris()

x = iris.data
y = iris.target

xtrain, xtest, ytrain, ytest = train_test_split(x, y, 
                                                test_size=0.2,
                                                random_state=100)

print("Xtrain shape :", xtrain.shape)
print("Xtest shape :", xtest.shape)
print("ytrain shape :", ytrain.shape)
print("ytest shape :", ytest.shape)
```

```python
from sklearn.linear_model import LogisticRegression

# 3. Initialize and train the model
reg=LogisticRegression()
reg.fit(xtrain,ytrain)
```

```python
xtest
```

```python
y_predict=reg.predict(xtest)
y_predict
```

```python
ytest
```

```python

# 4. Get user inputs
print("Please enter the four flower measurements:")
a = float(input("Sepal Length: "))
b = float(input("Sepal Width: "))
c = float(input("Petal Length: "))
d = float(input("Petal Width: "))

# 5. Predict
prediction = reg.predict([[a, b, c, d]])

# 6. Display the result
species = iris.target_names[prediction[0]]
print(f"\nThe model predicts that the species of this flower is: {species}")

```

```python
from sklearn.metrics import classification_report
print(classification_report(ytest,y_predict))
```

```python
# plot confusion matrix

from sklearn.metrics import confusion_matrix

cm=confusion_matrix(ytest,y_predict)
print(cm)
plt.figure(figsize=(10,10))
sns.heatmap(cm, annot=True, fmt='d')
plt.xlabel("Predicted")
plt.ylabel("Acual")
plt.title("Confusion Matrix")
plt.show()
```

```python
import plotly.express as px
from sklearn.metrics import confusion_matrix

# 1. Calculate the confusion matrix
cm = confusion_matrix(ytest, y_predict)

# 2. Create the heatmap using Plotly Express
fig = px.imshow(
    cm,
    text_auto=True,                # Displays the numbers inside the squares (replaces annot=True)
    color_continuous_scale='Blues', # Sets the color gradient (similar to Seaborn's default)
    labels=dict(x="Predicted", y="Actual", color="Count") # Labels for the hover tooltip
)

# 3. Customize layout and titles
fig.update_layout(
    title={
        'text': "Confusion Matrix",
        'x': 0.5,                  # Centers the title
        'xanchor': 'center'
    },
    width=600,                     # Defines the width (replaces figsize)
    height=600,                    # Defines the height
    xaxis_title="Predicted",
    yaxis_title="Actual"
)

# Optional: If you want to show class names (e.g., 'Cat', 'Dog') instead of 0, 1...
classes = ['Class 0', 'Class 1', 'Class 2']
fig.update_xaxes(tickvals=list(range(len(classes))), ticktext=classes)
fig.update_yaxes(tickvals=list(range(len(classes))), ticktext=classes)

# 4. Display the interactive plot
fig.show()
```
